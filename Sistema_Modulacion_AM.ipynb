# Sistema de Modulación en Amplitud (AM)
**Asignatura:** Señales y Sistemas  
**Entorno:** Python en Google Colab

Este cuaderno genera una señal moduladora y una portadora, implementa AM convencional y analiza la señal en tiempo y frecuencia. También simula ruido, distorsión por recorte y atenuación.

## 1. Importación de bibliotecas

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.fft import fft, fftfreq

plt.rcParams['figure.figsize'] = (12, 5)
plt.rcParams['axes.grid'] = True
print('Bibliotecas importadas correctamente.')
```

## 2. Definición de parámetros y generación de señales

```python
fs = 100_000       # Frecuencia de muestreo (Hz)
duracion = 0.02    # Duración de la simulación (s)
fm = 500           # Frecuencia de la señal moduladora (Hz)
fc = 5_000         # Frecuencia de la portadora (Hz)
Am = 1             # Amplitud de la señal moduladora
Ac = 1             # Amplitud de la portadora
mu = 0.7           # Índice de modulación (70 %)

t = np.arange(0, duracion, 1/fs)
mensaje = Am * np.cos(2 * np.pi * fm * t)
portadora = Ac * np.cos(2 * np.pi * fc * t)

print(f'Muestras generadas: {len(t)}')
```

## 3. Visualización de las señales de entrada

```python
plt.figure(figsize=(12, 6))
plt.subplot(2, 1, 1)
plt.plot(t, mensaje)
plt.title('Señal moduladora (información)')
plt.xlabel('Tiempo (s)')
plt.ylabel('Amplitud')

plt.subplot(2, 1, 2)
plt.plot(t, portadora)
plt.title('Señal portadora')
plt.xlabel('Tiempo (s)')
plt.ylabel('Amplitud')
plt.tight_layout()
plt.show()
```

## 4. Implementación de la modulación AM

```python
# AM convencional con portadora: s(t) = Ac[1 + mu*m(t)]cos(2*pi*fc*t)
senal_am = Ac * (1 + mu * mensaje) * np.cos(2 * np.pi * fc * t)

envolvente = Ac * (1 + mu * mensaje)
plt.figure(figsize=(12, 5))
plt.plot(t, senal_am, label='Señal AM')
plt.plot(t, envolvente, '--', label='Envolvente positiva')
plt.plot(t, -envolvente, '--', label='Envolvente negativa')
plt.title('Señal modulada en amplitud (AM)')
plt.xlabel('Tiempo (s)')
plt.ylabel('Amplitud')
plt.legend()
plt.tight_layout()
plt.show()
```

## 5. Análisis en el dominio de la frecuencia

```python
N = len(senal_am)
espectro = np.abs(fft(senal_am)) / N
frecuencias = fftfreq(N, d=1/fs)
indices = frecuencias >= 0

plt.figure(figsize=(12, 5))
plt.plot(frecuencias[indices], espectro[indices])
plt.title('Espectro de frecuencia de la señal AM')
plt.xlabel('Frecuencia (Hz)')
plt.ylabel('Magnitud')
plt.xlim(0, 10_000)
plt.tight_layout()
plt.show()

print('Componentes esperadas: banda lateral inferior =', fc-fm,
      'Hz; portadora =', fc, 'Hz; banda lateral superior =', fc+fm, 'Hz.')
```

## 6. Simulación de ruido gaussiano

```python
np.random.seed(42)  # Hace reproducible el ejemplo
snr_db = 10
potencia_senal = np.mean(senal_am**2)
potencia_ruido = potencia_senal / (10**(snr_db/10))
ruido = np.sqrt(potencia_ruido) * np.random.randn(len(t))
senal_con_ruido = senal_am + ruido

plt.figure(figsize=(12, 5))
plt.plot(t, senal_am, label='Señal AM original')
plt.plot(t, senal_con_ruido, label='Señal AM con ruido', alpha=0.7)
plt.title(f'Efecto del ruido en la señal AM (SNR = {snr_db} dB)')
plt.xlabel('Tiempo (s)')
plt.ylabel('Amplitud')
plt.legend()
plt.tight_layout()
plt.show()
```

## 7. Simulación de distorsión por recorte

```python
limite = 0.65
senal_distorsionada = np.clip(senal_am, -limite, limite)

plt.figure(figsize=(12, 5))
plt.plot(t, senal_am, label='Señal AM original')
plt.plot(t, senal_distorsionada, label='Señal distorsionada')
plt.title('Efecto de la distorsión por recorte')
plt.xlabel('Tiempo (s)')
plt.ylabel('Amplitud')
plt.legend()
plt.tight_layout()
plt.show()
```

## 8. Simulación de atenuación

```python
factor_atenuacion = 0.5
senal_atenuada = senal_am * factor_atenuacion

plt.figure(figsize=(12, 5))
plt.plot(t, senal_am, label='Señal AM original')
plt.plot(t, senal_atenuada, label='Señal AM atenuada')
plt.title('Efecto de la atenuación (50 %)')
plt.xlabel('Tiempo (s)')
plt.ylabel('Amplitud')
plt.legend()
plt.tight_layout()
plt.show()
```

## 9. Interpretación de resultados

- **Señal original:** presenta una envolvente que sigue la señal de información.
- **Ruido:** agrega variaciones aleatorias y dificulta observar la forma original.
- **Distorsión:** el recorte limita los picos y cambia la forma de onda.
- **Atenuación:** reduce la amplitud a la mitad sin cambiar idealmente las frecuencias.

Con los parámetros usados, se esperan componentes espectrales alrededor de 4,500 Hz, 5,000 Hz y 5,500 Hz. El índice de modulación es 0.7 (70 %), por lo que no hay sobremodulación en este ejemplo.
